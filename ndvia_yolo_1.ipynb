```python
# YOLOv8 설치
!pip install ultralytics -q

# 필요한 라이브러리 불러오기
import torch
from ultralytics import YOLO
import cv2
import matplotlib.pyplot as plt

# YOLOv8 모델 로드 (사전 학습된 COCO 데이터셋 기반)
model = YOLO('yolov8n.pt')  # 'yolov8s.pt', 'yolov8m.pt' 등으로 변경 가능

# 이미지 로드 (Google Colab 환경에서 사용자는 직접 업로드해야 함)
from google.colab import files
uploaded = files.upload()  # 업로드 창이 뜨면 분석할 이미지 선택

# 업로드된 이미지 파일명 가져오기
image_path = list(uploaded.keys())[0]

# 이미지 불러오기
image = cv2.imread(image_path)
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)  # OpenCV는 BGR로 읽으므로 RGB로 변환

# YOLO 모델로 객체 감지 수행
results = model(image)

# 사람(Person) 클래스 ID (COCO 데이터셋 기준: ID = 0)
person_count = sum(1 for r in results[0].boxes.cls if int(r) == 0)

# 결과 출력
print(f"감지된 사람 수: {person_count}")

# 감지된 결과 시각화
results[0].show()
```
